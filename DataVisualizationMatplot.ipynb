```python
#plt.legend use korle kon graph kon ta sheita show korbe
# adding labels using plt.legend()
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'] ,df['Programming'],label = 'Programming')
plt.plot(df['Year'] ,df['Digital Marketing'], label= 'Digital Marketing')

plt.legend(loc='best') #loc diye location bjhacche r by default best dile oye thik korbe koi bestposition
```

```python
#plt.legend use korle kon graph kon ta sheita show korbe
# adding labels using plt.legend()
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'] ,df['Programming'],label = 'Programming', color = 'black', linewidth = 5)
plt.plot(df['Year'] ,df['Digital Marketing'], label= 'Digital Marketing' , color = 'green' , linewidth = 4)

plt.legend(loc='best') #loc diye location bjhacche r by default best dile oye thik korbe koi bestposition
```

```python
# adding some visualization

plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'] ,df['Programming'],label = 'Programming', color ='#11b86f',linewidth=3,linestyle='dashed',marker='o',markersize=7)
plt.plot(df['Year'] ,df['Digital Marketing'], label= 'Digital Marketing', color='black',linewidth=3,linestyle='dashed' )

plt.legend()
```

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student Enrollment over 8 years')

plt.plot(df['Year'] ,df['Programming'],label = 'Programming', color ='#11b86f',linewidth=3,linestyle='dashed',marker='o',markersize=7)
plt.plot(df['Year'] ,df['Digital Marketing'], label= 'Digital Marketing', color='black',linewidth=3,linestyle='dashed' )

plt.legend()
plt.grid()

plt.show()
```

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# gender wise

gender_group = df.groupby('Gender').size()

gender_group.index
gender_group.values
```

```python
plt.bar(gender_group.index,gender_group.values,color=['pink','green'],edgecolor='black')

plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Bar Chart of Gender among students')
```

```python
#  Bar chart on real data

df = pd.read_csv('/content/student_data (1).csv')

df

```

```python
status = df.groupby('CompletionStatus').size()

plt.bar(status.index,status.values,color=['green','orange','red'],edgecolor='black')

plt.xlabel('Completion Status')
plt.ylabel('Count')
plt.title('Bar Chart of Course completion')
plt.show()
```

```python
# location

status = df.groupby('Location').size()

plt.bar(status.index,status.values,edgecolor='black')

plt.xlabel('Location')
plt.ylabel('Count')
plt.title('Bar Chart of Location of Students')
plt.show()
```

Pie chart

```python
status = df.groupby('CompletionStatus').size()
plt.pie(status,labels=status.index,autopct='%1.1f%%',explode = (0.1,0,0),shadow=True)
plt.show()
```
